# NumPy
NumPy short for Numerical Python, is one of the most important foundational packages for numerical computing in Python.  
Many computational packages providing scientific functionality use NumPy’s array objects as one of the standard interface lingua francas for data exchange.

## N-dimensional array

### Explicitly convert or cast an array from one data type to another using ndarray’s astype method

```python
arr_int = np.array([1, 2, 3], dtype=np.int32)
```

```python
arr_int.dtype
```

```python
arr_float = arr_int.astype(np.float64)
```

```python
arr_float.dtype
```
